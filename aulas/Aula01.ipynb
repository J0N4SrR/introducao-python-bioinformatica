### 1. O que é um Valor?

O **Valor** é a unidade básica de informação que um programa manipula. É o dado "nu e cru".

Na saúde, lidamos com valores o tempo todo:

* `"Positivo"` (um texto)
* `120` (um número inteiro de batimentos cardíacos)
* `5.2` (um número decimal de pH sanguíneo)

### 2. O que é um Objeto?

Em linguagens modernas como Python, dizemos que **"tudo é um objeto"**.

Pense no **Objeto** como uma cápsula que guarda três coisas:

1. **O Valor:** O dado em si (ex: `37`).
2. **O Tipo:** A categoria dele (ex: se é um número ou um texto).
3. **O Comportamento:** O que o computador tem permissão de fazer com ele (ex: números podem ser somados; textos podem ser escritos em letras maiúsculas).

## 🏷️ Variáveis: Dando Nome as coisas

Pense na rotina de um laboratório. O que acontece se você tem dez tubos idênticos com líquidos transparentes e nenhum deles tem uma etiqueta? 
O conteudo (valor) do o líquido está lá, mas ele é inútil porque você não sabe o que está.

Na programação, o erro é parecido. O computador guarda o valor na memória, mas se você não deu um nome (uma variável), você não consegue "chamar" esse dado de volta. 
Outro ponto importante: o sinal de igual (`=`) na programação não significa "igualdade matemática" (como em $2 + 2 = 4$), mas sim **atribuição**. É o ato de colocar o valor dentro do frasco e colar a etiqueta.

**Explicação do Conteúdo**
Uma **Variável** é, essencialmente, um nome que se refere a um valor. E, como o próprio nome diz, ela é **variável**: o conteúdo pode mudar ao longo do tempo, mas a etiqueta (o nome) permanece.

* **Atribuição Simples:** `dosagem = 50` (Você está dizendo: "Guarde o valor 50 na etiqueta chamada dosagem").
* **Atribuição Encadeada:** Às vezes, na pesquisa, vários grupos começam com a mesma condição inicial. O Python nos permite fazer isso de uma vez só:
`grupo_a = grupo_b = grupo_c = 0` (Todos começam com valor zero).



```python
#nome_da_variavel = valor

# Criando uma variável chamada dosagem
dosagem = 0

# Exibindo o valor armazenado
print(dosagem)
```

```python
# Valor inicial
# Atribuição de 50 a variavel dosagem
dosagem = 50

print("Dosagem inicial:", dosagem)

# Atualizando a variável
dosagem = 75

print("Nova dosagem:", dosagem)
```

```python
# Informações laboratoriais
paciente = "Ana"
temperatura = 36.8
resultado_covid = "Negativo"
quantidade_amostras = 12

# Exibindo os dados
print("Paciente:", paciente)
print("Temperatura:", temperatura)
print("Resultado:", resultado_covid)
print("Amostras:", quantidade_amostras)
```

```python
#Atribuição Encadeada
x = y = z = 6
print(x,y,z)

# Todos os grupos começam com zero
grupo_a = grupo_b = grupo_c = 0

print("Grupo A:", grupo_a)
print("Grupo B:", grupo_b)
print("Grupo C:", grupo_c)
```

### 🛠️ O que acontece "Debaixo do Capô"?

Com base no trecho que você enviou, aqui está a anatomia técnica de uma atribuição:

1. **O Identificador (O Nome):** É o rótulo que nós criamos (`x`, `y`, `nome`).
2. **O Endereço de Memória:** É a posição real nos "slots" de 32-bit (ou 64-bit) do sistema.
3. **O Conteúdo (O Valor):** É a informação guardada naquela posição específica.

### 🧬 Referências: O "Cabo de Guerra" entre Nomes e Objetos

Imagine que você tem um frasco com uma amostra e coloca uma etiqueta nela chamada `a`. Depois, você decide que a etiqueta `b` também deve apontar para essa mesma amostra. O que acontece se você decidir que a etiqueta `a` agora deve ser colada em um frasco totalmente novo? O frasco original (onde `b` ainda está colada) muda?

Muitos alunos acreditam que, ao fazer `b = a`, criamos um vínculo eterno entre as duas letras. Eles pensam: *"Se `b` veio de `a`, tudo o que acontecer com `a` deve refletir em `b`"*.

Mas pense na ética da pesquisa: se você tira uma foto (cópia) dos dados de um paciente hoje e, amanhã, o estado de saúde dele muda no hospital, a foto que você tirou ontem deveria mudar sozinha? **Não.** Na programação, para tipos básicos (números e textos), o comportamento é de independência.

---

### 🛠️ O que acontece na Memória (Passo a Passo)

Vamos seguir este roteiro de execução:

1. **`a = 10`**: O Python cria um objeto "10" na memória e a etiqueta `a` aponta para ele.
2. **`b = a`**: O Python diz: "Ok, `b` agora vai apontar para o **mesmo objeto** que `a` está apontando". Agora, os dois nomes olham para o número 10.
3. **`a = 20`**: Aqui está o segredo. O Python **não muda** o número 10 para 20. Ele cria um **novo objeto** "20" e move a etiqueta `a` para lá.

**Resultado Final:**

* `a` agora aponta para **20**.
* `b` continua apontando para o **10** original.


```python
print("ETAPA 1 -> a = 10")

# O Python cria o objeto 10
# e a variável 'a' aponta para ele
a = 10

print("Valor de a:", a)
print("ID de a:", id(a))
```

```python
print("\nETAPA 2 -> b = a")

# Agora 'b' aponta para o MESMO objeto de 'a'
b = a

print("Valor de a:", a)
print("Valor de b:", b)

print("ID de a:", id(a))
print("ID de b:", id(b))

# IDs iguais = mesmo objeto na memória
```

```python
print("\nETAPA 3 -> a = 20")

# O Python NÃO altera o objeto 10
# Ele cria um NOVO objeto 20
# e move apenas a referência de 'a'
a = 20

print("Valor de a:", a)
print("Valor de b:", b)

print("ID de a:", id(a))
print("ID de b:", id(b))


print("\nRESULTADO FINAL")

print(f"a aponta para -> {a}")
print(f"b aponta para -> {b}")
```

## **Explicação do Conteúdo**
Em Python, a variável é apenas um nome simbólico (um ponteiro). O dado de fato mora em um Objeto na memória do computador.

**A Variável:** É o seu "atalho", o nome que você digita no código para facilitar sua vida.

**O Objeto:** É o espaço físico na memória que contém o valor real (o número do batimento cardíaco, a sequência de DNA, o diagnóstico).

Quando você escreve glicemia = 90, você não está "enchendo" a palavra glicemia com o número 90. Você está criando o objeto 90 na memória e dizendo para a palavra glicemia apontar para ele.

```python
# Criamos um objeto com o resultado de um exame e damos um nome
resultado_exame = "Normal"

# Criamos uma segunda referência para o MESMO objeto (como um apelido)
laudo_final = resultado_exame

print(f"Referência 1 (resultado_exame): {resultado_exame}")
print(f"Referência 2 (laudo_final): {laudo_final}")

# O comando 'id()' nos mostra o "endereço do frasco". 
# Repare que os IDs serão IDÊNTICOS, pois o objeto é o mesmo:
print(f"Endereço na memória: {id(resultado_exame)}")
print(f"Endereço na memória: {id(laudo_final)}")
```

### **Números**

Em Python existem vários tipos de números, tais como `int` (integer), `float` (floating point) e `complex` (complex).


**1. Inteiros**

São números sem pontos decimais. No Python 3, efetivamente não há limite para o tamanho de um valor inteiro. Mas, ele é limitado pela quantidade de memória que seu sistema possui. Esses valores são do tipo `int`

```python
print(9187391837310731092037173)
```

```python
type(7136782163691941204)
```

Sem prefixo, o python interpreta a sequência de numero como base decimal.

Para indicar uma base diferente, você usa prefixos:

image.png

O tipo, independente da base usada, continua sendo `int`
