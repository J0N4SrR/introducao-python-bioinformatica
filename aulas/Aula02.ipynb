## 🧪 Estruturas Condicionais

### 👁️ O que é?
Estruturas condicionais permitem que algoritmos tomem decisões, executando blocos de código apenas se uma condição booleana (verdadeira/falsa) for atendida. 

Estruturas condicionais permitem que o programa **escolha caminhos diferentes com base nos dados**.

### 🏥 Analogia em Saúde: Triagem (Protocolo de Manchester)

Imagine a classificação de um paciente na triagem:

- Se a dor é **insuportável** → 🔴 Pulseira Vermelha  
- Se a dor é **leve** → 🟢 Pulseira Verde  
- Caso contrário → 🟡 Pulseira Amarela

### 🧠 Interpretação em programação
O computador faz exatamente isso:
- Testa uma condição  
- Se for verdadeira, executa uma ação  
- Caso contrário, tenta outra  
- Se nenhuma for atendida, segue o padrão

## Comparadores (Os Testes)

Antes de tomar decisões, o computador precisa **comparar valores**.  
O resultado de uma comparação é sempre:

- `True` (Verdadeiro)  
- `False` (Falso)

### 🔍 Operadores de Comparação

| Operador | Significado        | Exemplo Clínico                          |
|----------|------------------|------------------------------------------|
| `>`      | Maior que         | `temperatura > 37.5` (Febre?)            |
| `<`      | Menor que         | `glicemia < 70` (Hipoglicemia?)          |
| `==`     | Igual a           | `tipo_sanguineo == "O-"` (Doador universal?) |
| `!=`     | Diferente de      | `batimentos != 0` (Paciente vivo?)       |
| `>=`     | Maior ou igual    | `idade >= 60` (Grupo prioritário?)       |

### 🎯 Ideia principal
Cada comparação vira uma **pergunta de SIM ou NÃO**  
O Python responde com:
- `True` → Sim  
- `False` → Não  



````python
# ===============================
#  Dados do paciente
# ===============================

temperatura = 38.0        # temperatura corporal em graus Celsius
glicemia = 120            # nível de glicose no sangue
tipo_sanguineo = "O-"     # tipo sanguíneo do paciente
batimentos = 80           # batimentos cardíacos por minuto
idade = 70                # idade do paciente
````

````python
# ===============================
# 🔍 Testes (operações booleanas)
# ===============================

# Verificando se o paciente está com febre
# Se a temperatura for maior que 37.5 → True (febre)
print("Febre?", temperatura > 37.5)

# Verificando hipoglicemia
# Se a glicemia for menor que 70 → True (hipoglicemia)
print("Hipoglicemia?", glicemia < 70)

# Verificando tipo sanguíneo
# Compara se o tipo é exatamente "O-"
print("Doador universal?", tipo_sanguineo == "O-")

# Verificando sinais vitais
# Se batimentos forem diferentes de zero → paciente vivo
print("Paciente vivo?", batimentos != 0)

# Verificando grupo prioritário
# Se idade for maior ou igual a 60 → prioridade
print("Grupo prioritário?", idade >= 60)
````

## 🧪 Estrutura Simples: `if` (Se...)

### 🎯 O que é?
O `if` é uma estrutura condicional que permite **avaliar uma expressão** e, de acordo com o resultado, **executar uma ação**.

---

### 🚪 Como funciona?
O `if` é a **porta de entrada da decisão**:

- Se a condição for **verdadeira (`True`)** → o código é executado  
- Se for **falsa (`False`)** → o código é ignorado  

````python
# Verifica febre
if temperatura > 37.5:
    print("Paciente com febre")
````

````python
# Verifica hipoglicemia
if glicemia < 70:
    print("Paciente com hipoglicemia")
````

````python
# Verifica se está vivo (condição básica)
if batimentos != 0:
    print("Paciente com sinais vitais")
````

````python
# Verifica grupo prioritário
if idade >= 60:
    print("Paciente é grupo prioritário")
````

## 🧪 Estrutura Composta: `else` (Senão...)

### 🎯 O que é?
O `else` é usado quando queremos definir **o que acontece caso a condição do `if` NÃO seja atendida**.

---

### 🧠 Como funciona?
- `if` → executa se a condição for **True**  
- `else` → executa se a condição for **False**  

👉 Ou seja, o `else` é o **“Plano B”** da decisão.

---


````python
# Se a temperatura for maior que 37.5 → febre
if temperatura > 37.5:
    print("Paciente com febre")
else:
    print("Paciente sem febre")
````

````python
# Se glicemia for menor que 100 → normal
if glicemia < 100:
    print("Glicemia normal")
else:
    print("Glicemia alterada")
````

````python
# Se batimentos forem diferentes de zero → paciente vivo
if batimentos != 0:
    print("Paciente com sinais vitais")
else:
    print("Paciente sem sinais vitais")
````

````python
# Se idade >= 60 → prioridade
if idade >= 60:
    print("Paciente prioritário")
else:
    print("Paciente não prioritário")
````

## 🧪 Múltiplas Decisões: `elif` (Senão, se...)

### 🎯 O que é?
Na saúde, raramente temos apenas **“sim” ou “não”** — geralmente existem **níveis**.  
O `elif` permite testar **várias condições em sequência**.

---

### 🧠 Como funciona?
A estrutura segue esta lógica:

```text
if (condição 1):
    executa
elif (condição 2):
    executa
elif (condição 3):
    executa
else:
    executa (caso nenhuma condição seja verdadeira)

````python
glicemia = 120

if glicemia < 70:
    print("Hipoglicemia")
elif glicemia <= 99:
    print("Glicemia normal")
elif glicemia <= 125:
    print("Pré-diabetes")
else:
    print("Diabetes")
````

````python
temperatura = 38.0

if temperatura < 35:
    print("Hipotermia")
elif temperatura <= 37.5:
    print("Normal")
elif temperatura <= 39:
    print("Febre")
else:
    print("Febre alta")
````

````python
# Classificando frequência cardíaca
if batimentos < 60:
    print("Bradicardia (batimentos baixos)")
elif batimentos <= 100:
    print("Frequência cardíaca normal")
else:
    print("Taquicardia (batimentos elevados)")
````

````python
# Classificando faixa etária
if idade < 18:
    print("Menor de idade")
elif idade < 60:
    print("Adulto")
elif idade < 80:
    print("Idoso")
else:
    print("Idoso longevo")
````

# 🧠 Lógica Combinada em Python: `and` e `or`

Em saúde e ciência de dados, raramente uma decisão depende de apenas um fator isolado.  
Por isso, usamos **operações lógicas combinadas** para construir regras mais realistas.

---

# 🧾 Dados do paciente (caso clínico)

```python
# ===============================
# Paciente simulado
# ===============================

temperatura = 38.0        # °C
glicemia = 120            # mg/dL
batimentos = 80           # bpm
idade = 70                # anos
tipo_sanguineo = "O-"
```

---

# 🟢 Operador AND (E lógico)

## 📌 Ideia
Todas as condições precisam ser verdadeiras ao mesmo tempo.

> "Condição composta"

### 🧪 Exemplo clínico: febre + taquicardia

```python
risco_and = (temperatura > 37.5) and (batimentos > 100)

print("Risco combinado (AND):", risco_and)
```

### 🔎 Interpretação clínica

- Temperatura = 38.0 → ✔ febre
- Batimentos = 80 → ❌ não está elevado

➡ Resultado final: **False (sem risco combinado por AND)**

---

# 🔵 Operador OR (OU lógico)

## 📌 Ideia
Basta uma condição ser verdadeira.

> "Qualquer fator já aciona o alerta"

### 🧪 Exemplo clínico: fator de risco geral

```python
risco_or = (idade > 65) or (glicemia > 140)

print("Fator de risco (OR):", risco_or)
```

### 🔎 Interpretação clínica

- Idade = 70 → ✔ fator de risco
- Glicemia = 120 → ❌ normal

➡ Resultado final: **True (risco presente por idade)**

---

# 📊 Comparação rápida

| Operador | Regra lógica              |
|----------|--------------------------|
| AND      | Todas as condições       |        
| OR       | Pelo menos uma condição  |
