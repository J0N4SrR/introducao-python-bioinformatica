# 1. Manipulação de Strings 🔬

Em programação, chamamos qualquer texto de `String` (que significa "cordão" ou "sequência").

Uma `String` é simplesmente uma sequência de caracteres (letras, números ou símbolos) sempre escritos entre aspas (`" "`).

## Analogia Clínica

Pense em uma fita de DNA.

O DNA é uma sequência de nucleotídeos (`A`, `T`, `C`, `G`).

Para o Python, uma sequência de DNA é apenas uma `String`!

![Diagrama da Dupla Hélice do DNA e String Python](ImagemDNA.jpeg)

---

`````python
# Criando uma variável (uma "caixa") para guardar nossa sequência de DNA
sequencia_dna = "ATCGGTAA"

# O comando print() exibe o conteúdo na tela, como o resultado de um exame
print("A nossa sequência de DNA é:")
print(sequencia_dna)
`````

## O que aconteceu aqui?

Nós dissemos ao Python:

> "Pegue o texto `ATCGGTAA` e guarde dentro de um espaço chamado `sequencia_dna`."

Em seguida, pedimos para ele mostrar (`print`) o que estava lá dentro.

Simples assim!

## A Regra de Ouro do Python: O Índice Zero (0)

> Em Python, a indexação de sequências (listas, tuplas e strings) sempre começa em `0`.
>
> O primeiro elemento é `[0]`, o segundo é `[1]` e assim por diante.

---

## Índices em uma String

Quando temos uma `String`, cada letra ocupa uma posição, chamada de **Índice**.

- A primeira letra está na posição `0`
- A segunda letra está na posição `1`
- E assim por diante...

---

## Exemplo 

| Letra | A | T | C | G | G | T | A | A |
|---|---|---|---|---|---|---|---|---|
| Índice | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |

---

## Como acessar elementos

Usamos colchetes `[ ]` para acessar uma posição específica.

Exemplo:

```python
dna = "ATCG"

print(dna[0])  # Primeira letra
print(dna[1])  # Segunda letra
```

* `dna[0]` → `A`
* `dna[1]` → `T`

```
```


## Vamos testar na prática

Vamos extrair apenas a primeira base nitrogenada da nossa fita de DNA.


`````python
# Sequência de DNA original
dna = "ATGCGTAA"

# Extraindo a primeira base (posição 0) usando colchetes [ ]
primeira_base = dna[0]

# Extraindo a segunda base (posição 1)
segunda_base = dna[1]

print("A primeira base nitrogenada é:")
print(primeira_base)

print("A segunda base nitrogenada é:")
print(segunda_base)
`````

## O que aconteceu aqui?

Os colchetes `[ ]` funcionam como uma pinça de laboratório.

Ao escrever:

```python
dna[0]
````

você "pinçou" exatamente a letra que estava na posição zero (a letra `"A"`).


# Slicing (Fatiamento): Extraindo Códons ✂️

Na biologia, o DNA é lido em grupos de três nucleotídeos chamados códons.

Em Python, podemos fazer exatamente isso usando uma técnica chamada `Slicing` (fatiamento).

---

## O que é Slicing?

Slicing é uma forma de “recortar” apenas uma parte de uma sequência.

Funciona como uma tesoura digital ✂️

---

# Estrutura do Slicing

```python
sequencia[inicio:fim:passo]
````
---

## Como o Python interpreta isso?

### `inicio`

É onde o corte começa.

✅ Esse valor entra no recorte.

---

### `fim`

É onde o corte termina.

⚠️ O Python para ANTES desse número.

O valor final NÃO entra.

---

### `passo`

Define de quantos em quantos elementos o Python vai “pular”.

Esse valor é opcional.

---

# Exemplo Visual

```python
dna = "ATGCGTAA"
```

| Letra  | A | T | G | C | G | T | A | A |
| ------ | - | - | - | - | - | - | - | - |
| Índice | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |

---

## Vamos extrair o primeiro códon

Queremos pegar:

```python
ATG
```

Essas letras estão nas posições:

* `0`
* `1`
* `2`

Então fazemos:

```python
dna[0:3]
```

O Python:

* começa no índice `0`;
* percorre a sequência;
* para ANTES do índice `3`.

Resultado:

```python
ATG
```

```
```


`````python
# Sequência de DNA
dna = "ATGCGTAA"

# Extraindo o primeiro códon: começa no 0 e vai até o 3 (o 3 não é incluído, pega 0, 1 e 2)
primeiro_codon = dna[0:3]

print("O primeiro códon extraído é:")
print(primeiro_codon)

`````

`````python
# Fatiamento (Slicing)
# Estrutura: [inicio:fim:passo]
#
# inicio = onde começa
# fim = onde para (o valor final NÃO entra)
# passo = de quantos em quantos caracteres o Python vai pular
#
# dna[0::2]
# começa no índice 0
# vai até o final da sequência
# pulando de 2 em 2 caracteres

bases_pares = dna[0::2]

print("Resultado do fatiamento:")
print(bases_pares)
`````

# Ferramentas de Análise: Buscando e Contando  📊

O Python possui várias “ferramentas” embutidas para analisar textos e sequências.

Essas ferramentas são chamadas de **métodos**.

No nosso exemplo estamos trabalhando com sequencia de letras que representam bases nitrogenadas.
Tenha em mente que em uma analise real, nossa variavel teria milhares/milhoes de letras, assim, como strings podem representar sequências de DNA e RNA, esses métodos são extremamente úteis na análise de padrões genéticos.

---

## `.count()`

Conta quantas vezes um caractere ou sequência aparece.

### Exemplo clínico
Quantas Adeninas (`A`) existem em uma sequência de DNA?

---

## `.find()`

Procura a posição onde uma sequência começa.

### Exemplo clínico
Em qual posição começa um determinado padrão genético?

---

## `.replace()`

Substitui um trecho por outro.

### Exemplo clínico
Substituir Timina (`T`) por Uracila (`U`) para simular a transcrição do DNA em RNA.

`````python
# Sequência de DNA
dna = "ATGCGTAA"

# 1. Contando quantas bases Adenina (A) existem na fita
contagem_a = dna.count("A")
print("Quantidade de Adeninas na sequência:")
print(contagem_a)

# 2. Buscando onde começa o padrão "CGT"
posicao_padrao = dna.find("CGT")
print("O padrão 'CGT' começa no índice:")
print(posicao_padrao)

# 3. Transcrição básica: substituindo Timina (T) por Uracila (U)
rna = dna.replace("T", "U")
print("Sequência após substituição (DNA para RNA):")
print(rna)
`````

## O que aconteceu aqui?

O método:

```python
.count("A")
````

varreu toda a sequência e encontrou a letra `"A"` três vezes.

---

O método:

```python
.find("CGT")
```

retornou o número `2`, porque a sequência `"CGT"` começa exatamente no índice `2`.

---

O método:

```python
.replace("T", "U")
```

criou uma nova sequência de RNA, substituindo automaticamente todas as Timinas (`T`) por Uracilas (`U`).

Isso é extremamente útil para evitar erros manuais em sequências grandes.

```
```


`````python
# Texto simples
frase = "banana"

# 1. Contando quantas vezes a letra "a" aparece
contagem_a = frase.count("a")
print("Quantidade de letras 'a':")
print(contagem_a)

# 2. Procurando onde começa o trecho "nan"
posicao = frase.find("nan")
print("O trecho 'nan' começa no índice:")
print(posicao)

# 3. Substituindo letras
nova_frase = frase.replace("a", "o")
print("Texto após substituição:")
print(nova_frase)
`````

# Listas (Lists)

Uma lista funciona como uma coleção de informações guardadas em um único lugar.

Ela pode armazenar:
- nomes;
- números;
- exames;
- pacientes;
- medicamentos;
- ou qualquer outro tipo de dado.

As listas são criadas usando colchetes:

```python
[ ]
````

---

## Exemplo do cotidiano 🛒

Imagine uma lista de compras:

```python
compras = ["Arroz", "Leite", "Pão"]
```

Durante o dia você pode:

* adicionar novos itens;
* remover produtos;
* trocar informações.

A lista pode mudar o tempo todo.

---

# Característica Principal: Mutabilidade

A principal característica das listas é a **mutabilidade**.

Isso significa que podemos:

* alterar elementos;
* adicionar novos valores;
* remover informações;

mesmo com o programa funcionando.

---



`````python
# Criando uma lista de compras
compras = ["Arroz", "Leite", "Pão"]

print("Lista inicial de compras:")
print(compras)

# Adicionando um novo item à lista
compras.append("Café")

print("Lista de compras atualizada:")
print(compras)
`````

## Strings vs. Listas: O que muda na manipulação?

A diferença fundamental entre manipular Strings e manipular Listas em Python resume-se a dois conceitos técnicos: **Tipo de Conteúdo** e **Mutabilidade**.

### 1. O que armazenam?
*   **Strings:** Armazenam **exclusivamente texto** (uma sequência de caracteres contínuos).
*   **Listas:** Armazenam **qualquer tipo de dado** (números, textos, booleanos e até outras listas misturadas).

### 2. A Regra da Mutabilidade (Como operam na memória)
*   **Strings são IMUTÁVEIS:** Você não pode alterar uma letra isolada dentro de uma string já criada. Qualquer manipulação (como usar `.replace()` ou `.upper()`) não modifica a string original; em vez disso, o Python **cria e devolve uma nova string** com a alteração.
*   **Listas são MUTÁVEIS:** Você pode alterar, adicionar ou remover elementos da lista original a qualquer momento. Operações como `.append()` ou a substituição por índice modificam a estrutura **no mesmo espaço de memória**, sem criar uma cópia.

`````python
texto = "DNA"
texto[0] = "R" # ❌ TypeError! Strings não permitem trocar um item diretamente.
`````

`````python
# O correto é criar uma nova string:
novo_texto = texto.replace("D", "R")
print(novo_texto)
`````

`````python
exames = ["PCR", "Glicemia", "Colesterol"]
exames[0] = "Hemograma" # ✅ Funciona! O item na posição 0 foi substituído.

print(exames) 
# Resultado: ["Hemograma", "Glicemia", "Colesterol"]
`````

# Tuplas (Tuples)

As `Tuplas` são muito parecidas com as listas.

A principal diferença é que as tuplas são **imutáveis**.

Isso significa que, depois de criadas, elas não podem ser alteradas.

---

## Como as tuplas são criadas?

As tuplas usam parênteses:

```python
( )
````

---

# Característica Principal: Imutabilidade

Depois que uma tupla é criada:

* não podemos adicionar elementos;
* não podemos remover elementos;
* não podemos alterar valores internos.

Ela permanece fixa.

---

## Exemplo do cotidiano 🔒

Imagine o número do seu CPF.

Ele é uma informação fixa:

* você não adiciona números;
* não remove partes;
* não altera o valor durante o uso.

Algumas informações precisam permanecer estáveis e protegidas.

---

## Exemplo na Biomedicina 🧬

O códon de iniciação da tradução é:

```python
("A", "T", "G")
```

Se alterarmos essa estrutura, o processo biológico é comprometido.

Por isso, uma tupla é útil para representar dados que não devem mudar.



`````python
# Célula de Código

# Criando uma tupla representando um códon fixo
codon_inicio = ("A", "T", "G")

print("O códon de início é:")
print(codon_inicio)
`````

`````python

# Tuplas não podem ser modificadas
# O código abaixo geraria erro:
codon_inicio.append("C")
`````

# O que aconteceu aqui?

A tupla protegeu a estrutura da informação.

O Python entende que esse dado deve permanecer fixo.

Por isso:

* não podemos adicionar elementos;
* não podemos remover itens;
* não podemos modificar a estrutura.

Tuplas são úteis para armazenar informações importantes que precisam permanecer seguras durante a execução do programa.

# Dicionários (Dictionaries)

Os `Dicionários` organizam informações em pares de:

- Chave (`Key`)
- Valor (`Value`)

Eles são criados usando chaves:

```python
{ }
````

---

## Como funciona?

Em listas, buscamos informações pela posição:

```python
lista[0]
```

Nos dicionários, buscamos informações pelo nome da chave.

---

## Exemplo do cotidiano 📒

Imagine a agenda do seu celular.

Você não procura um contato pela posição da lista.

Você procura pelo nome da pessoa.

O nome funciona como a chave.

O telefone funciona como o valor.

---

## Exemplo na Biomedicina 🧬

Podemos usar um dicionário para associar:

* Nome de um gene;
* Código da proteína correspondente.

Funciona como um pequeno banco de dados.

`````python
# Criando um dicionário com genes e códigos UniProt
genes = {
    "SPG4": "Q9UBP0",
    "BRCA1": "P38398",
    "TP53": "P04637"
}

print("Banco de dados de genes:")
print(genes)

# Buscando o valor associado à chave "BRCA1"
codigo_brca1 = genes["BRCA1"]

print("Código associado ao gene BRCA1:")
print(codigo_brca1)
`````

# O que aconteceu aqui?

O dicionário armazenou informações em formato de:

```python
chave : valor
````

Quando escrevemos:

```python
genes["BRCA1"]
```

o Python procurou a chave `"BRCA1"` e devolveu o valor associado a ela:

```python
P38398
```

Isso é muito útil para organizar:

* dados clínicos;
* pacientes;
* exames;
* genes;
* resultados laboratoriais.



`````python
# Criando um dicionário com informações de um paciente
paciente = {
    "ID_Paciente": "Maria",
    "Exame": "Hemograma"
}

# Mostrando o dicionário completo
print("Dados do paciente:")
print(paciente)

# Acessando apenas o nome do paciente pela chave
print("Nome da paciente:")
print(paciente["ID_Paciente"])

# Acessando o exame solicitado
print("Exame solicitado:")
print(paciente["Exame"])
`````

`````python
album = {'nome':'A Night at the Opera','artista':'Blind Guardian','lançamento':2002}
print(type(album)) # <class 'dict'>
print(album) # {'nome': 'A Night at the Opera', 'artista': 'Blind Guardian', 'lançamento': 2002}
print(album['nome']) # A Night at the Opera
print(album['artista']) # Blind Guardian
print(album['lançamento']) # 2002
`````
