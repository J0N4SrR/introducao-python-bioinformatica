# 🔁 3. Automatizando Tarefas com Loops

---

## 🧠 O Problema

Imagine analisar manualmente centenas de exames de glicemia.

Seria necessário:
- pegar um exame;
- verificar o valor;
- decidir se existe alerta;
- repetir isso várias vezes.

Esse tipo de tarefa é repetitiva.

Na programação, utilizamos os **loops** para automatizar esse processo.

---

## 🔄 O que é um Loop?

Um loop permite repetir ações automaticamente.

O `for` percorre vários elementos de uma lista, um por vez.

![O que aconteceu no código?](Comparação_Loops.png)

---

## 📱 Analogia do Feed

Em redes sociais, o algoritmo faz algo parecido:

- pega um post;
- analisa;
- decide se será mostrado;
- passa para o próximo.

O loop funciona da mesma forma:
ele percorre cada item da lista automaticamente.

---


`````python
# Importando a biblioteca de tempo
import time

print("--- INICIANDO RONDA DE TRIAGEM ---")

# O loop percorre cada paciente da lista
for p in pacientes:

    # Pequena pausa de 1 segundo
    time.sleep(1)

    # Verificando a glicemia
    if p["glicemia"] > 99:

        print(f"⚠️ ALERTA: {p['nome']} está com glicemia alterada!")

    else:

        print(f"✅ {p['nome']}: Glicemia normal.")

print("--- TRIAGEM CONCLUÍDA ---")
`````

`````python
# Importando a biblioteca de tempo
import time

indice = 0

print("--- MONITORAMENTO INICIADO ---")

# O while continuará executando
# enquanto a glicemia for menor que 180
while pacientes[indice]["glicemia"] < 180:

    # Exibindo o índice atual
    print(f"📍 Índice atual: {indice}")

    # Exibindo o nome do paciente
    print(f"✅ {pacientes[indice]['nome']} sem risco crítico.")

    # Pausa de 1 segundo
    time.sleep(1)

    # Passando para o próximo paciente
    indice += 1

print("\n \n🚨 PACIENTE CRÍTICO ENCONTRADO!")
print(f"🚨 {pacientes[indice]['nome']} precisa de Atenção")
`````

# 🔁 `break` e `continue` em Loops

---

## ⛔ `break`

O `break` é utilizado para **interromper completamente a execução de um loop**.

Quando ele é executado:
- o loop é encerrado imediatamente;
- nenhuma repetição adicional acontece;
- o programa continua a execução normalmente após o loop.

👉 Em outras palavras, o `break` “força a saída” do loop antes dele terminar naturalmente.

---

## ⏭️ `continue`

O `continue` é utilizado para **pular a iteração atual do loop**.

Quando ele é executado:
- o restante do código dentro do loop naquela repetição é ignorado;
- o fluxo passa diretamente para a próxima iteração;
- o loop continua normalmente.

👉 Em outras palavras, o `continue` “pula uma rodada” do loop sem encerrá-lo.

---

## 🧠 Diferença principal

- `break` → encerra o loop completamente  
- `continue` → pula apenas a repetição atual e segue o loop

## 🧩 A Ferramenta de Organização: Funções (`def`)



Em um laboratório ou ambiente clínico, ninguém reinventa o processo de análise toda vez que precisa medir uma glicemia. Existe um **POP (Procedimento Operacional Padrão)**, que garante consistência, segurança e eficiência.


Na programação, o equivalente ao POP são as **funções**.

Uma função é uma forma de **definir uma tarefa uma única vez** e depois apenas **reutilizar esse comportamento sempre que necessário**, chamando pelo nome.

---

## 🧩 Estrutura básica (pseudocódigo)

```
def nome_da_funcao(parametros):
    fazer algo
    retornar resultado
```

---

## 🐍 Estrutura real em Python

```python
def nome_da_funcao(parametro1, parametro2):
    # lógica da função
    resultado = parametro1 + parametro2
    return resultado
```

---

## 🧪 Exemplo simples

### ➕ Função de soma



`````python
def somar(a, b):
    resultado = a + b
    return resultado 
`````

`````python
#Usando a função
total = somar(5, 3)
print(total)
`````

`````python
# Protocolo de triagem glicêmica
def triagem_glicemia(valor):
    """
    Recebe um valor de glicemia e retorna a conduta clínica adequada.
    """

    if valor >= 250:
        return "🚨 URGENTE: Hiperglicemia grave → Encaminhar imediatamente para emergência / avaliar cetose / hidratação venosa 🚨"

    elif valor >= 150:
        return " 🔴 ALTO RISCO: Encaminhar para endocrinologista + ajustar conduta medicamentosa + avaliar HbA1c"

    elif valor >= 100:
        return "🟡 ATENÇÃO: Glicemia alterada → Solicitar Curva Glicêmica + orientação alimentar + reavaliação clínica"

    elif valor >= 70:
        return "🟢 NORMAL: Paciente estável → Manter acompanhamento de rotina"

    else:
        return "🟠 ALERTA: Hipoglicemia → Orientar ingestão imediata de carboidrato + reavaliar + investigar causa"
`````

`````python
resultado = triagem_glicemia(120)
print(resultado)
`````

`````python
import time

# 🔁 Usando função dentro de repetição (loop)
for p in pacientes:
    conduta = triagem_glicemia(p["glicemia"])    
    time.sleep(0.5)
    print(f"👤 {p['nome']} - {conduta}")
    print("-" * 150)

    time.sleep(1)  # pausa entre pacientes
`````
