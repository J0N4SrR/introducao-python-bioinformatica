## ⚙️ Os quatro superpoderes do Pandas

O Pandas foi desenvolvido para lidar com dados do mundo real e se apoia em quatro grandes capacidades:

### 🧹 Limpeza (Cleaning)

Permite identificar e tratar dados ausentes ou inconsistentes.

Facilita a análise inicial dos dados, permitindo obter rapidamente medidas como média, valores máximos e variabilidade da amostra com poucos comandos.

Exemplo: cruzar resultados de exames laboratoriais com registros de dispensação farmacêutica usando o ID do paciente.

### 📊 Análise (Analyzing)

Transforma dados brutos em conhecimento útil, permitindo gerar insights a partir das informações integradas.


# 📊 DataFrame

---

No dia a dia, ninguém analisa um único valor isolado. Nós trabalhamos com listas de pacientes, catálogos de medicamentos e planilhas lotadas de exames. Precisamos de uma estrutura prática para organizar tudo isso em formato de grade.

---

### O que é, afinal, um DataFrame?

De forma simples e direta: o **DataFrame** é a versão do Pandas para uma tabela comum.

Se você sabe o que é uma planilha do Excel/Sheets, você já entende o que é um DataFrame. Ele é uma estrutura bidimensional composta por:

* **Linhas (`Index`):** Representam as nossas *observações*. Cada linha é um Paciente, uma Amostra de Sangue ou uma Molécula diferente.
* **Colunas (`Columns`):** Representam as nossas *variáveis*. Cada coluna guarda uma informação, como a Idade, a Glicemia ou o Peso Molecular.

**Por que usar o DataFrame e não o Excel?**

Porque o DataFrame processa dados via código. Isso significa que ele não trava ao abrir milhões de linhas e permite que você filtre, limpe e faça cálculos estatísticos complexos em segundos.

`````python
# Aqui usamos o NumPy para gerar uma lista de números (idades)
# np.array cria um vetor numérico eficiente

idades = np.array([18, 22, 25, 30, 35])
# Também criamos outro vetor com notas (simulando um exame ou avaliação)
notas = [7.5, 8.0, 6.5, 9.0, 7.0]

animais = {
    "Animal": ["Cachorro", "Gato", "Elefante", "Leão", "Coelho"],
    "Habitat": ["Doméstico", "Doméstico", "Savana", "Savana", "Floresta"],
    "Expectativa_de_vida": [13, 15, 60, 14, 9],
    "Peso_medio_kg": [10, 4, 5400, 190, 2]
}

print("Idades dos alunos:", idades)
print("Resultados (notas):", notas)
print("Dicionario de animais:", animais)
`````

`````python
# ================================
# 📊 Construindo a tabela
# ================================
# Aqui estamos pegando dados que já foram criados antes:
# - idades 
# - notas
# E organizando tudo em colunas nomeadas.

df = pd.DataFrame({
    "Idade": idades,  # coluna com as idades
    "Nota": notas     # coluna com as notas/resultados
})

df_animais = pd.DataFrame(animais)
`````

`````python
# Agora imprimimos o DataFrame para ver como os dados ficaram organizados

print("📊 Tabela de dados (DataFrame):")
print(df)
`````

`````python

# Exibimos a tabela animais
print("\nTabela de Animais:")
print(df_animais)
`````

O Pandas cria DataFrames a partir de diferentes estruturas: listas viram colunas simples quando passadas isoladamente; arrays NumPy são convertidos diretamente em colunas eficientes para cálculos; já dicionários são o formato mais natural, onde cada chave vira uma coluna e os valores viram as linhas. Isso facilita a organização e análise de dados em formato de tabela.

## 🚀 Como o Pandas transforma dados

### 📋 Lista → vira uma coluna quando agrupada em um DataFrame  
### ⚡ NumPy array → funciona como lista, mas otimizado para cálculos  
### 🗂️ Dicionário → já nasce estruturado em colunas (chave = coluna, valores = linhas)

👉 No final, o Pandas pega tudo isso e organiza em uma tabela chamada DataFrame.

# 🧼 Valores Nulos no Pandas

Em ciência de dados, é muito comum encontrar **dados faltando** em uma tabela.  
Esses valores ausentes precisam ser identificados e tratados antes da análise.

---

## ❌ O que é NaN?

- `NaN` significa **"Not a Number"**
- Representa um **dado faltante ou não registrado**
- Pode aparecer por:
  - falha na coleta de dados
  - campos não preenchidos
  - erro de importação de arquivos

---


`````python
# ================================
# 📂 Lendo um arquivo Excel
# ================================
# Aqui estamos carregando um arquivo pronto (.xlsx)
# Esse arquivo pode conter dados incompletos (valores faltando)
df_nulos =  pd.read_excel("../data/mock_dados_valores_nulos.xlsx")
`````

`````python
#`isna()` → identificar valores faltantes
#A função `isna()` verifica onde existem valores ausentes.
df_nulos.isna()
`````

`````python
#Podemos combinar com sum() para saber quantos valores faltam em cada coluna:
# mostra a quantidade de NaN por coluna
df_nulos.isna().sum()
`````

`````python
# ================================
# 🧹 dropna() -> remove valores nulos
# ================================

# Remove linhas com valores nulos
df_drop = df_nulos.dropna()

# Remove linhas onde "Salario" está vazio
df_salario = df_nulos.dropna(subset=["Salario"])

# Remove colunas que possuem valores nulos
df_colunas = df_nulos.dropna(axis=1)
 
# ================================
# 📊 Exibindo resultados
# ================================

print("\n🧹 Linhas sem valores nulos:")
print(df_drop)

print("\n💰 Linhas com salário preenchido:")
print(df_salario)


print("\n📋 Colunas sem valores nulos:")
print(df_colunas)
`````

`````python
# ================================
# 🔧 Preenchendo valores nulos
# ================================
# Substitui valores faltantes por um valor escolhido

df_preenchido = df_nulos.fillna({
    "Idade": df["Idade"].mean(),
    "Cidade": "Desconhecido",
    "Profissão": "Não informado",
    "Salario": df["Salario"].median()
})

print(df_preenchido)
`````

`````python
# ================================
# 🔄 astype() -> converte tipos de dados
# ================================

# Verificando os tipos atuais
print(df_nulos.dtypes)
`````

`````python
# Convertendo a coluna Idade para inteiro

df_preenchido["Idade"] = df_preenchido["Idade"].astype(int)

# Convertendo salário para float
df_preenchido["Salario"] = df_preenchido["Salario"].astype(float)

# Convertendo Nome para texto/string
df_preenchido["Nome"] = df_preenchido["Nome"].astype(str)


# ================================
# 📊 Verificando tipos das colunas
# ================================

print("📌 Tipo da coluna 'Idade' antes do astype():")
print(df_nulos["Idade"].dtype)

print("\n🔄 Tipo da coluna 'Idade' depois do astype(int):")
print(df_preenchido["Idade"].dtype)

print("\n📋 DataFrame final:")
print(df_preenchido)
`````

## 🔄 Ordenação no Pandas: `sort_values()`

O método `sort_values()` é usado para **ordenar linhas** de um DataFrame com base em uma ou mais colunas.

---

## ⚡ Parâmetros mais usados

| Parâmetro | Função |
|---|---|
| `by` | coluna usada na ordenação |
| `ascending=True` | crescente |
| `ascending=False` | decrescente |
| `inplace=True` | altera o DataFrame original |


`````python
# 📌 Sintaxe básica

df.sort_values(by="Idade")
`````

`````python
# 🔽 Ordem decrescente
df.sort_values(by="Salario", ascending=False)
`````

`````python
# 🧩 Ordenando por mais de uma coluna
df.sort_values(by=["Cidade", "Salario"])
`````

## 📌 Diferença entre `sort_values()` e `sort_index()`

| Método | O que ordena |
|---|---|
| `sort_values()` | valores das colunas |
| `sort_index()` | índices/linhas |

---

## 🚀 Resumo mental rápido

📊 `sort_values()` → ordena pelos valores da coluna  
🔽 `ascending=False` → maior para menor  
🧩 `by=[]` → múltiplas colunas  
🛠️ `inplace=True` → altera o DataFrame original

`````python
dados = {
    "Nome": ["Ana", "Carlos", "Bruno"],
    "Idade": [23, 35, 19]
}

df = pd.DataFrame(dados, index=[2, 0, 1])

print("📋 DataFrame original:")
print(df)

print("\n🔢 Índice ordenado:")
print(df.sort_index())
`````

## 📊 `value_counts()` no Pandas

O método `value_counts()` conta quantas vezes cada valor aparece em uma coluna.

`````python
# Conta quantas vezes cada cidade aparece
contagem_cidades = df_nulos["Cidade"].value_counts()

print("📍 Quantidade de pessoas por cidade:")
print(contagem_cidades)
`````

# 📊 Agregações com `groupby()` no Pandas

O método `groupby()` serve para **agrupar dados** por uma coluna e depois aplicar cálculos (média, soma, contagem, etc).

👉 Muito usado para análises e relatórios.

---

## 🧠 Estrutura mental do `groupby()`

```python
df.groupby("COLUNA")[["COLUNA_NUMERICA"]].METODO()
```

---
## 🔧 Métodos de agregação mais usados

| Método    | Função      |
| --------- | ----------- |
| `mean()`  | Média       |
| `sum()`   | Soma        |
| `count()` | Contagem    |
| `max()`   | Maior valor |
| `min()`   | Menor valor |

---

### 👉 Uma forma MUITO mais fácil de entender `groupby()` é ler o código **de trás pra frente**.

---

## 📌 Exemplo

```python
df.groupby("Cidade")["Salario"].mean()
```

["Salario"]

👉 agrupando por cidade


`````python
# ================================
# 📊 MÉDIA SALARIAL POR CIDADE
# ================================

# 1. Agrupa os dados pela coluna "Cidade"
# 2. Seleciona a coluna "Salario"
# 3. Calcula a média dos salários

media_salario = df_preenchido.groupby("Cidade")["Salario"].mean().round(2)

print("📊 Média salarial por cidade:")
print(media_salario)
`````

`````python
# ================================
# 🔢 QUANTIDADE DE PESSOAS POR PROFISSÃO
# ================================

# Conta quantos nomes existem em cada profissão

contagem_profissao = df_preenchido.groupby("Profissão")["Nome"].count()

print("\n👨‍💼 Quantidade de pessoas por profissão:")
print(contagem_profissao)
`````

`````python
# ================================
# 💰 SOMA DOS SALÁRIOS POR CIDADE
# ================================

# Soma todos os salários de cada cidade

soma_salarios = df_preenchido.groupby("Cidade")["Salario"].sum()

print("\n💰 Soma dos salários por cidade:")
print(soma_salarios)
`````

`````python
# ================================
# MAIOR IDADE POR CIDADE
# ================================

# Mostra a maior idade encontrada em cada cidade

maior_idade = df_preenchido.groupby("Cidade")["Idade"].max()

print("\n Maior idade por cidade:")
print(maior_idade)
`````

`````python
# ⬇️ MENOR SALÁRIO POR PROFISSÃO
# ================================

# Mostra o menor salário de cada profissão

menor_salario = df_preenchido.groupby("Profissão")["Salario"].min()

print("\n Menor salário por profissão:")
print(menor_salario)
`````

# 📚 Agrupando por mais de uma coluna no `groupby()`

No Pandas, podemos passar uma **lista de colunas** dentro do `groupby()`.

👉 Isso cria grupos mais específicos.

---

## 📌 Estrutura

```python
df.groupby(["COLUNA_1", "COLUNA_2"])["COLUNA_NUMERICA"].METODO()

`````python
# ================================
# 📊 AGRUPANDO POR CIDADE E PROFISSÃO
# ================================

# O groupby() vai criar grupos usando:
# 1. Cidade
# 2. Profissão
#
# Depois:
# - selecionamos a coluna "Salario"
# - calculamos a média salarial de cada grupo

media_salario = df_preenchido.groupby(["Cidade", "Profissão"])["Salario"].mean().round(2)

print("📊 Média salarial por cidade e profissão:")
print(media_salario)
`````

`````python
# ================================
# 👥 CONTANDO PESSOAS POR CIDADE E STATUS
# ================================

# Agrupa por:
# - Nome
# - Cidade
# - Ativo
#
# Depois agrupa ela media das idades

contagem = df_preenchido.groupby(["Nome", "Cidade", "Ativo"])["Idade"].mean().round(2)

print("\n👥 Quantidade de pessoas por cidade e ativo:")
print(contagem)
`````

# 📊 `.agg()` no Pandas

O método `.agg()` permite aplicar **várias agregações ao mesmo tempo**.

👉 Em vez de fazer apenas média OU soma, podemos fazer tudo junto.

---

# 📌 Estrutura

```python
df.groupby("COLUNA")["COLUNA_NUMERICA"].agg(["mean", "max", "min"])
```

---

`````python
# ================================
# 📊 MÚLTIPLAS AGREGAÇÕES
# ================================

# Agrupa os dados pela coluna "Cidade"
# Depois calcula:
# - média
# - maior valor
# - menor valor

estatisticas = df_preenchido.groupby(
    "Cidade"
)["Salario"].agg([
    "mean",
    "max",
    "min"
])

print("📊 Estatísticas salariais por cidade:")
print(estatisticas)
`````

`````python
estat_cidade = df_preenchido.groupby("Cidade").agg({
    "Salario": ["mean", "max"],
    "Idade": ["min", "max"]
})

print("📊 Estatísticas por cidade:")
estat_cidade
`````
