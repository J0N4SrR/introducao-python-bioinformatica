## 🧪 Questão 04 – O Cálculo do IMC (Potência e Parênteses)

### 🎯 Objetivo
Entender a ordem de importância nas operações matemáticas.

### 📝 Tarefa
Calcule o IMC de um paciente com **80 kg** e **1.75 m** de altura.

### 🧠 Lógica
Fórmula: **IMC = Peso / (Altura²)**  
Em Python: `peso / (altura * altura)`

````python
# Dados do paciente
peso = 80
altura = 1.75

# Calculando o IMC: Peso dividido por (Altura ao quadrado)
imc = peso / (altura ** 2)

# Exibindo sem arredondamento 
print("O IMC do paciente é:", imc)
# Exibindo com arredondamento (2 casas decimais)
print("O IMC do paciente é:", round(imc, 2))
````

## 🧪 Questão 05 – Conversão de Exames (Casting)

### 🎯 Objetivo
Transformar um dado que veio como texto em um número.

### 📝 Tarefa
Crie `glicemia_texto = '120'`.  
Converta para inteiro usando `int()` e some 10 unidades ao valor.

### 🧠 Lógica
Matemática: transformação de tipo para permitir soma aritmética.

````python
# Exemplo de Erro – Somar sem Converter (Casting)

# O dado chegou como texto (entre aspas)
glicemia_texto = "120"

# Tentando somar diretamente (isso vai dar erro)
resultado_final = glicemia_texto + 10

print("Glicemia corrigida:", resultado_final)
````

````python
# O dado chegou como texto (entre aspas)
glicemia_texto = "120"

# Convertendo para número inteiro para permitir cálculos
glicemia_numero = int(glicemia_texto)

# Agora podemos somar unidades
resultado_final = glicemia_numero + 10

print("Glicemia corrigida:", resultado_final)
````
