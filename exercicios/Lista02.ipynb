# 1. Alerta de Febre (O comando `if`)

## Objetivo
Identificar uma condição simples.

## Tarefa
Crie uma variável `temperatura = 38.2`.

Se o valor for maior que `37.5`, exiba a mensagem:

```python
'Paciente com febre'

`````python
# Passo 1: Definimos a temperatura do paciente e guardamos na variável
temperatura = 38.2

# Passo 2: Usamos o comando 'if' para verificar a condição
# Lemos assim: "SE a temperatura for maior (>) que 37.5, então faça o seguinte:"
if temperatura > 37.5:
    
    # Passo 3: Esta linha só será executada se a condição acima for Verdadeira (True)
    # O espaço inicial (indentação) avisa ao Python que esta ação pertence ao 'if'
    print("Paciente com febre")
`````

# 2. Sinais Vitais (O comando `else`)

## Objetivo
Decidir entre dois caminhos (Normal vs. Alterado).

## Tarefa
Crie uma variável `batimentos = 110`.

Se for maior que `100`, exiba:

```python
'Taquicardia'
```

Caso contrário (else), exiba:

```python
'Frequência Normal'
```

Dica: O else é o seu plano B para quando o if é falso.

`````python
# Criamos a variável com a frequência cardíaca do paciente
batimentos = 110

# Testamos a primeira condição (Plano A)
# Lemos: "SE os batimentos forem maiores que 100:"
if batimentos > 100:
    # Se for verdade, executa esta linha
    print("Taquicardia")

# Se a condição acima for falsa, o Python pula para cá (Plano B)
# Lemos: "CASO CONTRÁRIO:"
else:
    # Se entrou no else, significa que os batimentos são 100 ou menos
    print("Frequência Normal")
`````

# 3. Triagem de Manchester (O comando `elif`)

## Objetivo
Trabalhar com múltiplas faixas de decisão.

## Tarefa
Crie uma variável `nivel_dor` (de 0 a 10).

- Se for de `0 a 3`, exiba:

```python
'Verde'
```

Se for de 4 a 7, exiba:
```python
'Amarelo'
```

Se for 8 ou mais, exiba:

```python
'Vermelho'
```

Lógica: Use elif para testar as faixas intermediárias.


`````python
# Registramos o nível de dor relatado pelo paciente (escala de 0 a 10)
nivel_dor = 5

# Teste 1: Dor leve
# Lemos: "SE o nível de dor for menor ou igual a 3:"
if nivel_dor <= 3:
    print("Cor da Triagem: Verde")

# Teste 2: Dor moderada
# Lemos: "SE NÃO FOI O ANTERIOR, SE o nível de dor for menor ou igual a 7:"
# (Sabemos que é maior que 3, pois o primeiro 'if' já falhou para chegar aqui)
elif nivel_dor <= 7:
    print("Cor da Triagem: Amarelo")

# Teste 3: Dor intensa (Plano final)
# Lemos: "CASO CONTRÁRIO de todos os testes anteriores:"
else:
    print("Cor da Triagem: Vermelho")
`````

# 4. Grupo de Risco Vacinal (Operador lógico `AND`)

## Objetivo
Verificar se dois critérios são verdadeiros ao mesmo tempo.

## Tarefa
Crie:

```python
idade = 65
comorbidade = True
````

Se a idade for maior que `60` **E** `comorbidade` for `True`, exiba:

```python
'Grupo prioritário'
```

## Lógica

```python
(Condição A) and (Condição B)
```


`````python
# Definimos os dados do paciente
idade = 65
# A variável 'comorbidade' é um Booleano (Verdadeiro ou Falso)
comorbidade = True

# Usamos o operador lógico 'and' para exigir que ambas as regras sejam cumpridas
# Lemos: "SE a idade for maior que 60 E a comorbidade for Verdadeira:"
if idade > 60 and comorbidade == True:
    
    # Esta linha só imprime se as DUAS afirmações acima forem verdades
    print("Grupo prioritário")
`````

# 5. Alerta de Emergência (Operador lógico `OR`)

## Objetivo
Decidir se pelo menos um critério de risco está presente.

## Tarefa
Crie:

```python
dor_peito = False
falta_ar = True
````

Se `dor_peito` for `True` **OU** `falta_ar` for `True`, exiba:

```python
'Alerta: Encaminhar para Emergência'
```

## Lógica

Basta uma condição ser verdadeira para o alerta disparar.

```
```


`````python
# Definimos o quadro clínico do paciente
dor_peito = False
falta_ar = True

# Usamos o operador lógico 'or' para criar uma condição abrangente
# Lemos: "SE o paciente tiver dor no peito OU SE tiver falta de ar:"
if dor_peito == True or falta_ar == True:
    
    # Basta uma das variáveis ser True para emitir o alerta
    print("Alerta: Encaminhar para Emergência")
`````

# 7. O Planejador de Domingo

## Objetivo
Aprender a traduzir um desenho (fluxograma) para a lógica do computador.

## Cenário
Você criou um assistente pessoal para decidir o que fazer no seu dia de folga.

Para funcionar, ele precisa checar três coisas:

- se há sol;
- se há dinheiro;
- se você tem energia.

## Sua Tarefa
Olhe para o fluxograma abaixo e "escreva" o que ele está dizendo em Python.

---

## Fluxograma

![Fluxograma - Planejador de Domingo](diagram2.png)

---


## Siga este roteiro

### 1. Crie os sensores
Defina as variáveis:

```python
esta_sol
tem_dinheiro
tem_energia
````

com `True` ou `False` (fica à sua escolha).

---

### 2. Siga as setas

* Se estiver `SOL` **E** tiver `DINHEIRO`, o destino é:

```python
'Praia'
```

* Se `NÃO` tiver `SOL` **OU** estiver `SEM ENERGIA`, o destino é:

```python
'TV'
```

* Se não for nenhuma das opções anteriores, o destino é:

```python
'Cinema'
```

```
```


`````python
# Passo 1: Criando os sensores (nossas variáveis de entrada)
# Vamos simular um domingo onde não há sol, mas temos dinheiro e energia.
esta_sol = False
tem_dinheiro = True
tem_energia = True

# Passo 2: Traduzindo o primeiro losango do fluxograma
# Caminho SIM da primeira pergunta
if esta_sol == True and tem_dinheiro == True:
    print("Resultado: Vou à Praia")

# Passo 3: Traduzindo o segundo losango (caso a primeira seta seja NÃO)
# Usamos 'elif' para fazer o segundo teste.
# Checamos se é falso que tem sol OU se é falso que tem energia
elif esta_sol == False or tem_energia == False:
    print("Resultado: Fico em casa assistindo TV")

# Passo 4: O último caminho restante do fluxograma
# Se não atendeu a nada acima, vai para a resposta final (Cinema)
else:
    print("Resultado: Vou ao Cinema")
`````
